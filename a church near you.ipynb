```python
import pandas as pd
import requests

# Official Church of England ArcGIS Feature Service for active churches
BASE_URL = (
    "https://services5.arcgis.com/KDRjxGRQDVgVtFTS/ArcGIS/rest/services/"
    "Active_churches_polygon_data/FeatureServer/0/query"
)

all_records = []
offset = 0
chunk_size = 2000

while True:
    params = {
        "where": "1=1",
        "outFields": "*",
        "resultOffset": offset,
        "resultRecordCount": chunk_size,
        "f": "json",
    }
    resp = requests.get(BASE_URL, params=params, timeout=30)
    data = resp.json()

    features = data.get("features", [])
    if not features:
        break

    for f in features:
        all_records.append(f.get("attributes", {}))

    print(f"Fetched {len(all_records)} churches...")
    if not data.get("exceededTransferLimit"):
        break

    offset += chunk_size

churches_df = pd.DataFrame(all_records)
churches_df.to_csv("cof_e_active_churches.csv", index=False)
print(f"Total churches saved: {len(churches_df)}")
```
